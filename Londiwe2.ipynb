# **Part B - Sets**

```python
#Exercise 7
set_favourite_subject = set()
print(type(set_favourite_subject))
```

```python
#Exercise 8
favourite_subject = {"Maths", "English", "Maths", "Science", "Life Orientation", "English", "Maths", "Science"}
print(favourite_subject)
#Due to elimination of duplicates, after printing only 4 distinct subjected were left [from 8 subjects mentioned by students].
```

```python
#Exercise 9
print(sorted(favourite_subject))
```

```python
#Exercise 10
favourite_subject.add("Geography")
print(favourite_subject)
favourite_subject.add("Maths")
print(favourite_subject)
#A set eliminates duplicates, hence adding Maths will make no difference since in the items in the set Maths is already there.
```

```python
#Exercise 11
favourite_subject.remove("Life Orientation")
print(sorted(favourite_subject))
```

```python
#Exercise 12
print("Science" in favourite_subject)
print("History" in favourite_subject)
```

```python
#Exercise 13
Monday = {"Sipho", "Aisha", "Thabo", "Naledi", "Pieter"}
Wednesday = {"Thabo", "Zanele", "Aisha", "Fatima", "Pieter"}
print(sorted(Monday.intersection(Wednesday)))
```
